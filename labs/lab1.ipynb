# Cell 4: a random variable is a function

A loan officer is about to open one of four applicant files, equally likely: A, B, C, D.
Let $X$ be the number of late payments on the file.

| file | X |
|------|---|
| A    | 0 |
| B    | 0 |
| C    | 1 |
| D    | 3 |

$X$ is not the file. $X$ is a function from those four outcomes to a number.

1. Write the distribution of $X$: $P(X=0)$, $P(X=1)$, $P(X=3)$.
   
   ANSWER: $P(X=0) = 0.5$, $P(X=1) = 0.25$, $P(X=3) = 0.25$

2. What is $P(X \ge 1)$? Use the complement.

    ANSWER: $P(X \ge 1) = 1 - P(X<1)$

    $P(X \ge 1) = 1 - 0.5$

    $P(X \ge 1) = 0.5$

```python
p_x0 = 0.5
p_x1 = 0.25
p_x3 = 0.25
p_at_least_one = 1 - p_x0
vals = [p_x0, p_x1, p_x3, p_at_least_one]
if all(v is not None for v in vals):
    assert abs(p_x0 - 1/2) < 1e-12
    assert abs(p_x1 - 1/4) < 1e-12
    assert abs(p_x3 - 1/4) < 1e-12
    assert abs(p_at_least_one - 1/2) < 1e-12
    assert abs(p_x0 + p_x1 + p_x3 - 1) < 1e-12
p_x0, p_x1, p_x3, p_at_least_one
```
