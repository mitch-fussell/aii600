# Cell 4: a random variable is a function

A loan officer is about to open one of four applicant files, equally likely: A, B, C, D.
Let $X$ be the number of late payments on the file.

| file | X |
|------|---|
| A    | 0 |
| B    | 0 |
| C    | 1 |
| D    | 3 |

$X$ is not the file. $X$ is a function from those four outcomes to a number.

1. Write the distribution of $X$: $P(X=0)$, $P(X=1)$, $P(X=3)$.
   
   **ANSWER: $P(X=0) = 0.5$, $P(X=1) = 0.25$, $P(X=3) = 0.25$**

2. What is $P(X \ge 1)$? Use the complement.

    **ANSWER: $P(X \ge 1) = 1 - P(X<1)$**

    **$P(X \ge 1) = 1 - 0.5$**

    **$P(X \ge 1) = 0.5$**

```python
p_x0 = 0.5
p_x1 = 0.25
p_x3 = 0.25
p_at_least_one = 1 - p_x0
vals = [p_x0, p_x1, p_x3, p_at_least_one]
if all(v is not None for v in vals):
    assert abs(p_x0 - 1/2) < 1e-12
    assert abs(p_x1 - 1/4) < 1e-12
    assert abs(p_x3 - 1/4) < 1e-12
    assert abs(p_at_least_one - 1/2) < 1e-12
    assert abs(p_x0 + p_x1 + p_x3 - 1) < 1e-12
p_x0, p_x1, p_x3, p_at_least_one
```

## Cell 5: at least one, and overlap

A barcode scanner fails to read 8% of packages, independently. Two packages come down the belt.

1. What is $P(\text{at least one failure})$? Use independence and the complement. do not add $0.08 + 0.08$

**ANSWER:**

**$P(\text{package reads correctly}) = 1 - 0.08 = 0.92$**
   
   **Therefore: $P(\text{package reads correctly for both}) = 0.92 * 0.92 = 0.8464$**
   
   **Therefore: $P(\text{at least 1 failure}) = 1 - P(\text{no failures}) = 1 - 0.8464 = 0.1536$**



2. Estimate it with a simulation of at least 10,000 trials

**ANSWER**
**running the 2 package test $100,000$ times an recording if at least on of the packages fails read will be a different value than working out the probabilty. However, the more simulations you run, the closer it will get to that probabilty score.**

  
A second, smaller question, using the overlap rule from the slides. Tonight's email goes to people who opened last week $(P(A) = 0.12)$ or who clicked last week $(P(B) = 0.05)$. Both: $P(A \text{ and } B) = 0.03$. They are not independent: $P(A)P(B) = 0.006$, not $0.03$. What is $P(A \text{ or } B)$?

**ANSWER: $P(A \text{ or } B) = P(A) + P(B) - P(A \text{ and } B)$**

**Therefore: $P(A \text{ or } B) = 0.12 + 0.05 - 0.03 = 0.14$**

```python
import random
p_fail = 0.08
p_at_least = 0.1536
random.seed(1)

n = 100000
count = 0

for _ in range(n):
    package1 = random.random() < p_fail
    package2 = random.random() < p_fail
    if package1 or package2:
        count = count + 1
p_sim = count/n

p_open_or_click = 0.14

print(p_sim)
```

## Cell 6: independence twice

A fraud filter scores three independent transactions a night. Each is flagged with probability $0.2$ A night is noisy if at least one transaction is flagged.

1. What is $P(a noisy night)$? Complement, then independence within the night.

**ANSWER**
$P(\text{One transaction not flagged}) = 1 - 0.2 = 0.8$

$P(\text{No transactions flagged}) = 0..8 * 0.8 0.8 = 0.512$

$P(\text{Noisy night}) = 1 - 0.512 = 0.488$

2. What is $P(five noisy nights in a row)$? Independence across nights.

$P(\text{Five noisy nights in a row}) = 0.488 * 0.488 * 0.488 * 0.488 * 0.488 = 0.277 $

This is the hitting-streak move from class: a probability built inside one block, then raised to the number of blocks. Do not use Rose’s batting average or 44 games.

```python
p_noisy = 0.488
p_five = 0.488**5
if p_noisy is not None and p_five is not None:
    assert abs(p_noisy - (1 - 0.8**3)) < 1e-12
    assert abs(p_five - (1 - 0.8**3)**5) < 1e-12
p_noisy, p_five
```
